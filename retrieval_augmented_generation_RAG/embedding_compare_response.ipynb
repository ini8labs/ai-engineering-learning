```python
# Import libraries
from sentence_transformers import SentenceTransformer, util
import numpy as np
```

```python
# Load a pre-trained model
model = SentenceTransformer('all-MiniLM-L6-v2')  # fast & good quality
```

```python
# Define your texts
texts = [
    "I love machine learning.",
    "AI is fascinating.",
    "I enjoy studying deep learning.",
    "The sky is blue today."
]
```

```python
# Generate embeddings
embeddings = model.encode(texts)
```

```python
# Compare embeddings
# Let's compute cosine similarity between all pairs
similarity_matrix = util.cos_sim(embeddings, embeddings)
print("Cosine Similarity Matrix:")
print(similarity_matrix)
```

```python
# Compare two specific sentences
sentence1 = "I love machine learning."
sentence2 = "I enjoy studying deep learning."
embedding1 = model.encode(sentence1)
embedding2 = model.encode(sentence2)

cos_sim = util.cos_sim(embedding1, embedding2)
print(f"\nCosine similarity between:\n'{sentence1}'\nand\n'{sentence2}'\n= {cos_sim.item():.4f}")
```
