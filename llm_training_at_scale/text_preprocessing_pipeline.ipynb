```python
# Import
import re
import hashlib
import unicodedata
from collections import Counter
import nltk
```

```python
# Safe download (avoids LookupError)
try:
    nltk.data.find('tokenizers/punkt')
except LookupError:
    nltk.download('punkt')

from nltk.tokenize import word_tokenize
```

```python
# Sample data
texts = [
    "Hello!!! This is a SAMPLE text",
    "Hello!!! This is a SAMPLE text",  # duplicate
    "Buy now!!! Visit http://spam.com",
    "Short text",
    "This is a clean and meaningful sentence for training purposes.",
    "Machine learning models require high quality datasets for better performance.",
    "Natural language processing is an exciting field of artificial intelligence."
]
```

```python
# Text cleaning
def clean_text(text):
    text = text.lower()
    text = re.sub(r"http\S+", "", text)  # remove URLs
    text = unicodedata.normalize('NFKD', text).encode('ascii', 'ignore').decode('utf-8')  # remove emojis
    text = re.sub(r"[^a-z0-9\s]", "", text)  # remove special chars
    text = re.sub(r"\s+", " ", text).strip()
    return text

cleaned_texts = [clean_text(t) for t in texts]

print("Cleaned Texts:")
print(cleaned_texts)
```

```python
# Deduplication
def remove_duplicates(texts):
    seen = set()
    unique_texts = []

    for text in texts:
        hash_val = hashlib.md5(text.encode()).hexdigest()
        if hash_val not in seen:
            seen.add(hash_val)
            unique_texts.append(text)

    return unique_texts

dedup_texts = remove_duplicates(cleaned_texts)

print("\n After Deduplication:")
print(dedup_texts)
```

```python
# Quality filtering
def is_high_quality(text, min_length=5):
    words = text.split()

    # Rule 1: Minimum length
    if len(words) < min_length:
        return False

    # Rule 2: Avoid repetition
    word_counts = Counter(words)
    most_common_word, freq = word_counts.most_common(1)[0]

    if freq / len(words) > 0.5:
        return False

    return True

filtered_texts = [t for t in dedup_texts if is_high_quality(t)]

print("\n After Quality Filtering:")
print(filtered_texts)
```

```python
# Tokenization
def tokenize_texts(texts):
    try:
        return [word_tokenize(text) for text in texts]
    except LookupError:
        print("NLTK issue detected, using simple split instead")
        return [text.split() for text in texts]

tokenized_texts = tokenize_texts(filtered_texts)

print("\n Tokenized Output:")
for tokens in tokenized_texts:
    print(tokens)
```

```python
# Final output
print("\n Final Processed Data:")
print(tokenized_texts)
```
