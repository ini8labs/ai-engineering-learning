```python
### Implement the MNIST neural network( using numpy) from scratch and experiment with different learning rates (0.001, 0.01, 0.1, 1.0) ###
```

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist
```

```python
# Load dataset
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

```python
# preprocess data

X_train = X_train.reshape(60000, 784) / 255
X_test = X_test.reshape(10000, 784) / 255
```

```python
# One-hot encoding
y_train_oh = np.eye(10)[y_train]
```

```python
# Activation functions
def relu(x):
    return np.maximum(0, x)

def softmax(x):
    exp = np.exp(x)
    return exp / np.sum(exp, axis=1, keepdims=True)
```

```python
# Training function
def train(lr):

    # Initialize weights
    W1 = np.random.randn(784, 64) * 0.01
    W2 = np.random.randn(64, 10) * 0.01

    losses = []

    for epoch in range(10):

        # Forward pass
        Z1 = X_train @ W1
        A1 = relu(Z1)

        Z2 = A1 @ W2
        A2 = softmax(Z2)

        # Loss
        loss = -np.mean(np.sum(y_train_oh * np.log(A2 + 1e-8), axis=1))
        losses.append(loss)

        # Backpropagation
        dZ2 = A2 - y_train_oh
        dW2 = A1.T @ dZ2 / len(X_train)

        dA1 = dZ2 @ W2.T
        dZ1 = dA1 * (Z1 > 0)
        dW1 = X_train.T @ dZ1 / len(X_train)

        # Update weights
        W1 -= lr * dW1
        W2 -= lr * dW2

        print(f"LR {lr} | Epoch {epoch+1} | Loss {loss:.4f}")

    return losses
```

```python
# Learning rates
learning_rates = [0.001, 0.01, 0.1, 1.0]

results = {}

for lr in learning_rates:
    results[lr] = train(lr)
    print(results)
```

```python
# Plot results
for lr in learning_rates:
    plt.plot(results[lr], label=f"lr={lr}")

plt.title("Learning Rate Comparison")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.show()
```
