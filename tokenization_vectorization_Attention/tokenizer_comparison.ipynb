```python
# Import library

import tiktoken
```

```python
# Example text
text = """
Large Language Models are transforming AI.
Tokenization is the process of converting text into tokens
that models like GPT can understand.
"""
```

```python
# Load tokenizers
gpt2_tokenizer = tiktoken.get_encoding("r50k_base")      # GPT-2 tokenizer
gpt4_tokenizer = tiktoken.get_encoding("cl100k_base")    # GPT-4 tokenizer
```

```python
# Encode text
tokens_gpt2 = gpt2_tokenizer.encode(text)
tokens_gpt4 = gpt4_tokenizer.encode(text)
```

```python
# Print token counts
print("GPT-2 tokenizer (r50k_base) token count:", len(tokens_gpt2))
print("GPT-4 tokenizer (cl100k_base) token count:", len(tokens_gpt4))
```

```python
# Show tokens
print("\nGPT-2 tokens:")
print(tokens_gpt2)

print("\nGPT-4 tokens:")
print(tokens_gpt4)
```

```python
# Compare
if len(tokens_gpt2) > len(tokens_gpt4):
    print("\nGPT-4 tokenizer produces fewer tokens.")
elif len(tokens_gpt2) < len(tokens_gpt4):
    print("\nGPT-2 tokenizer produces fewer tokens.")
else:
    print("\nBoth produce the same number of tokens.")
```

```python
''' The cl100k_base tokenizer (used in GPT-4) produces fewer tokens than r50k_base (used in GPT-2).

GPT-4’s tokenizer has a larger vocabulary (~100K tokens).

Because of this, it can represent bigger words or subwords as a single token.

Therefore, the text is split into fewer tokens, giving better compression. '''
```
