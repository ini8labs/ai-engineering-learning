```python
# Import libraries
import gensim
import gensim.downloader as api
from gensim.models import Word2Vec
```

```python
# Load a large corpus
dataset = api.load("text8")   # cleaned Wikipedia text
```

```python
# Train word2vec model
model = Word2Vec(
    sentences=dataset,
    vector_size=100,
    window=5,
    min_count=5,
    workers=4
)
```

```python
# Test analogies ( Example 1)
model.wv.most_similar(positive=["king", "woman"], negative=["man"], topn=3)
```

```python
# Example 2 (country and capital)
model.wv.most_similar(positive=["paris", "italy"], negative=["france"], topn=3)
```

```python
# Example 3 (verb tense)
model.wv.most_similar(positive=["walking", "swim"], negative=["walk"], topn=3)
```

```python
# Example 4 (gender relationship)
model.wv.most_similar(positive=["actor", "woman"], negative=["man"], topn=3)
```

```python
# Example analogy queries
queries = [
    (["king","woman"],["man"]),
    (["paris","italy"],["france"]),
    (["walking","swim"],["walk"]),
    (["bigger","small"],["big"]),
]

for pos, neg in queries:
    print(pos, "-", neg)
    print(model.wv.most_similar(positive=pos, negative=neg, topn=3))
    print()
```
