```python
import numpy as np
from tensorflow.keras.datasets import mnist
```

```python
# load dataset

(X_train, y_train), (X_test, y_test) = mnist.load_data()

X_train = X_train.reshape(60000, 784) / 255
X_test = X_test.reshape(10000, 784) / 255

y_train_oh = np.eye(10)[y_train]
```

```python
# activations functions

def relu(x):
    return np.maximum(0, x)

def softmax(x):
    exp = np.exp(x - np.max(x, axis=1, keepdims=True))
    return exp / np.sum(exp, axis=1, keepdims=True)
```

```python
# accuracy functions

def accuracy(pred, labels):
    pred_classes = np.argmax(pred, axis=1)
    return np.mean(pred_classes == labels)
```

```python
# MODEL 1 (1 hidden layer)
# 784 → 64 → 10

def train_model1(lr=0.01, epochs=200):

    # He initialization
    W1 = np.random.randn(784,64) * np.sqrt(2/784)
    W2 = np.random.randn(64,10) * np.sqrt(2/64)

    # Bias
    b1 = np.zeros((1,64))
    b2 = np.zeros((1,10))

    for epoch in range(epochs):

        # Forward
        Z1 = X_train @ W1 + b1
        A1 = relu(Z1)

        Z2 = A1 @ W2 + b2
        A2 = softmax(Z2)

        loss = -np.mean(np.sum(y_train_oh * np.log(A2 + 1e-8), axis=1))

        # Backprop
        dZ2 = A2 - y_train_oh
        dW2 = A1.T @ dZ2 / len(X_train)
        db2 = np.mean(dZ2, axis=0, keepdims=True)

        dA1 = dZ2 @ W2.T
        dZ1 = dA1 * (Z1 > 0)
        dW1 = X_train.T @ dZ1 / len(X_train)
        db1 = np.mean(dZ1, axis=0, keepdims=True)

        # Update
        W1 -= lr * dW1
        W2 -= lr * dW2
        b1 -= lr * db1
        b2 -= lr * db2

        if epoch % 10 == 0:
            print(f"Model1 Epoch {epoch} Loss {loss:.4f}")

    # Test accuracy
    A1_test = relu(X_test @ W1 + b1)
    A2_test = softmax(A1_test @ W2 + b2)

    return accuracy(A2_test, y_test)
```

```python
# MODEL 2 (3 hidden layers)
# 784 → 128 → 64 → 32 → 10

def train_model2(lr=0.01, epochs=200):

    W1 = np.random.randn(784,128) * np.sqrt(2/784)
    W2 = np.random.randn(128,64) * np.sqrt(2/128)
    W3 = np.random.randn(64,32) * np.sqrt(2/64)
    W4 = np.random.randn(32,10) * np.sqrt(2/32)

    b1 = np.zeros((1,128))
    b2 = np.zeros((1,64))
    b3 = np.zeros((1,32))
    b4 = np.zeros((1,10))

    for epoch in range(epochs):

        # Forward
        Z1 = X_train @ W1 + b1
        A1 = relu(Z1)

        Z2 = A1 @ W2 + b2
        A2 = relu(Z2)

        Z3 = A2 @ W3 + b3
        A3 = relu(Z3)

        Z4 = A3 @ W4 + b4
        A4 = softmax(Z4)

        loss = -np.mean(np.sum(y_train_oh * np.log(A4 + 1e-8), axis=1))

        # Backprop
        dZ4 = A4 - y_train_oh
        dW4 = A3.T @ dZ4 / len(X_train)
        db4 = np.mean(dZ4, axis=0, keepdims=True)

        dA3 = dZ4 @ W4.T
        dZ3 = dA3 * (Z3 > 0)
        dW3 = A2.T @ dZ3 / len(X_train)
        db3 = np.mean(dZ3, axis=0, keepdims=True)

        dA2 = dZ3 @ W3.T
        dZ2 = dA2 * (Z2 > 0)
        dW2 = A1.T @ dZ2 / len(X_train)
        db2 = np.mean(dZ2, axis=0, keepdims=True)

        dA1 = dZ2 @ W2.T
        dZ1 = dA1 * (Z1 > 0)
        dW1 = X_train.T @ dZ1 / len(X_train)
        db1 = np.mean(dZ1, axis=0, keepdims=True)

        # Update
        W1 -= lr * dW1
        W2 -= lr * dW2
        W3 -= lr * dW3
        W4 -= lr * dW4

        b1 -= lr * db1
        b2 -= lr * db2
        b3 -= lr * db3
        b4 -= lr * db4

        if epoch % 10 == 0:
            print(f"Model2 Epoch {epoch} Loss {loss:.4f}")

    # Test accuracy
    A1 = relu(X_test @ W1 + b1)
    A2 = relu(A1 @ W2 + b2)
    A3 = relu(A2 @ W3 + b3)
    A4 = softmax(A3 @ W4 + b4)

    return accuracy(A4, y_test)
```

```python
# Train & Compare

acc1 = train_model1()
acc2 = train_model2()

print("\nFinal Accuracy Comparison")
print("-------------------------")
print(f"Model 1 (1 hidden layer): {acc1:.4f}")
print(f"Model 2 (3 hidden layers): {acc2:.4f}")
```
